```python
# ============================================================
# Experiment: Basic Python Operations and TensorFlow 2 Tensors
# Generative AI Laboratory
# ============================================================

# Step 1: Import Required Libraries
import tensorflow as tf
import numpy as np

# Step 2: Verify TensorFlow Installation
print("TensorFlow Version:", tf.__version__)

# ============================================================
# PART A: BASIC PYTHON OPERATIONS
# ============================================================

print("\n========== BASIC PYTHON OPERATIONS ==========\n")

# Variables
a = 20
b = 10

print("Value of a:", a)
print("Value of b:", b)

# Arithmetic Operations
print("\nArithmetic Operations:")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)

# List Operations
numbers = [10, 20, 30, 40, 50]

print("\nList Elements:")
for num in numbers:
    print(num)

# Function Example
def square(x):
    return x * x

print("\nSquare of 5:", square(5))

# ============================================================
# PART B: TENSOR CREATION
# ============================================================

print("\n========== TENSOR CREATION ==========\n")

# 0-D Tensor (Scalar)
scalar = tf.constant(100)

print("Scalar Tensor:")
print(scalar)

# 1-D Tensor (Vector)
vector = tf.constant([10, 20, 30, 40])

print("\nVector Tensor:")
print(vector)

# 2-D Tensor (Matrix)
matrix = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print("\nMatrix Tensor:")
print(matrix)

# 3-D Tensor
tensor3d = tf.constant([
    [[1,2], [3,4]],
    [[5,6], [7,8]]
])

print("\n3-D Tensor:")
print(tensor3d)

# ============================================================
# PART C: TENSOR PROPERTIES
# ============================================================

print("\n========== TENSOR PROPERTIES ==========\n")

print("Scalar Shape:", scalar.shape)
print("Scalar Rank:", tf.rank(scalar))
print("Scalar Data Type:", scalar.dtype)

print("\nVector Shape:", vector.shape)
print("Vector Rank:", tf.rank(vector))
print("Vector Data Type:", vector.dtype)

print("\nMatrix Shape:", matrix.shape)
print("Matrix Rank:", tf.rank(matrix))
print("Matrix Data Type:", matrix.dtype)

print("\n3D Tensor Shape:", tensor3d.shape)
print("3D Tensor Rank:", tf.rank(tensor3d))
print("3D Tensor Data Type:", tensor3d.dtype)

# ============================================================
# PART D: TENSOR ARITHMETIC OPERATIONS
# ============================================================

print("\n========== TENSOR ARITHMETIC ==========\n")

tensor1 = tf.constant([10, 20, 30])
tensor2 = tf.constant([5, 10, 15])

print("Tensor 1:", tensor1)
print("Tensor 2:", tensor2)

# Addition
add = tf.add(tensor1, tensor2)
print("\nAddition:", add)

# Subtraction
sub = tf.subtract(tensor1, tensor2)
print("Subtraction:", sub)

# Multiplication
mul = tf.multiply(tensor1, tensor2)
print("Multiplication:", mul)

# Division
div = tf.divide(tensor1, tensor2)
print("Division:", div)

# ============================================================
# PART E: MATHEMATICAL FUNCTIONS
# ============================================================

print("\n========== MATHEMATICAL FUNCTIONS ==========\n")

data = tf.constant([4.0, 9.0, 16.0, 25.0])

print("Original Tensor:", data)

# Square
print("Square:")
print(tf.square(data))

# Square Root
print("Square Root:")
print(tf.sqrt(data))

# Mean
print("Mean:")
print(tf.reduce_mean(data))

# Maximum
print("Maximum:")
print(tf.reduce_max(data))

# Minimum
print("Minimum:")
print(tf.reduce_min(data))

# Sum
print("Sum:")
print(tf.reduce_sum(data))

# ============================================================
# PART F: TENSOR RESHAPING
# ============================================================

print("\n========== TENSOR RESHAPING ==========\n")

tensor = tf.constant([
    [1,2,3],
    [4,5,6]
])

print("Original Tensor:")
print(tensor)

# Reshape to 3x2
reshaped = tf.reshape(tensor, [3,2])

print("\nReshaped Tensor (3x2):")
print(reshaped)

# Flatten
flatten = tf.reshape(tensor, [-1])

print("\nFlatten Tensor:")
print(flatten)

# ============================================================
# PART G: SPECIAL TENSORS
# ============================================================

print("\n========== SPECIAL TENSORS ==========\n")

# Zeros Tensor
zeros = tf.zeros([3,3])

print("Zeros Tensor:")
print(zeros)

# Ones Tensor
ones = tf.ones([2,4])

print("\nOnes Tensor:")
print(ones)

# Random Tensor
random_tensor = tf.random.uniform([2,3])

print("\nRandom Tensor:")
print(random_tensor)

# Identity Matrix
identity = tf.eye(4)

print("\nIdentity Matrix:")
print(identity)

# ============================================================
# PART H: TENSOR DIMENSION OPERATIONS
# ============================================================

print("\n========== DIMENSION OPERATIONS ==========\n")

tensor = tf.constant([1,2,3,4])

print("Original Tensor:")
print(tensor)

# Expand Dimensions
expanded = tf.expand_dims(tensor, axis=0)

print("\nExpanded Tensor:")
print(expanded)

# Squeeze
squeezed = tf.squeeze(expanded)

print("\nSqueezed Tensor:")
print(squeezed)

# ============================================================
# PART I: TENSOR TO NUMPY CONVERSION
# ============================================================

print("\n========== TENSOR TO NUMPY ==========\n")

tensor = tf.constant([10,20,30,40])

numpy_array = tensor.numpy()

print("Tensor:")
print(tensor)

print("\nConverted NumPy Array:")
print(numpy_array)

print("Type:", type(numpy_array))

# ============================================================
# PART J: MATRIX MULTIPLICATION
# ============================================================

print("\n========== MATRIX MULTIPLICATION ==========\n")

A = tf.constant([
    [1,2],
    [3,4]
])

B = tf.constant([
    [5,6],
    [7,8]
])

result = tf.matmul(A, B)

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nMatrix Multiplication Result:")
print(result)

# ============================================================
# PART K: SIMPLE GENERATIVE AI EXAMPLE
# ============================================================

print("\n========== GENERATIVE AI EXAMPLE ==========\n")

# Simulated image data (4x4 pixels)
image = tf.random.uniform([4,4], minval=0, maxval=255, dtype=tf.int32)

print("Generated Image Tensor:")
print(image)

print("\nImage Shape:", image.shape)

# Normalize Image
normalized = tf.cast(image, tf.float32) / 255.0

print("\nNormalized Image:")
print(normalized)

# ============================================================
# END OF EXPERIMENT
# ============================================================

print("\nExperiment Completed Successfully.")
```
